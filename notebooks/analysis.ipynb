```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
names = []
bias_mean_salary = []

bias_mean_salary = pd.DataFrame({'name':names,'bias_mean_salary':bias_mean_salary})

bias_mean_salary.hist(bins=10)
```

```python
bias_mean_salary['bias_mean_salary'].mean()
```

```python
salary = []

salary = pd.DataFrame({'salary':salary})
```

```python
fig = plt.figure(figsize=(16,6))
ax1 = fig.add_subplot(121)
bias_mean_salary.hist(ax=ax1)
ax2 = fig.add_subplot(122)
salary.hist(ax=ax2)
plt.legend()
```

```python
mean_salary = salary.mean().iloc[0]
print("Year mean gross salary : €", mean_salary)
```

```python
bias_mean_salary.assign(error= lambda x: abs(x['bias_mean_salary']-mean_salary)).sort_values(by=['error'])
```
